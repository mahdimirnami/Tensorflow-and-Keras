# **import library data california**

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

# **Convert to Object**

```python
housing = fetch_california_housing()
```

# **train and test Model**

```python
X_train_full , X_test , Y_train_full , Y_test = train_test_split(housing.data,housing.target)
X_train ,X_valid , Y_train , Y_valid = train_test_split(X_train_full , Y_train_full)
```

# **Classifier with Keras**

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
X_test = scaler.transform(X_test)
```

# **model concatenate with one input and one output**

```python
              #       ┌─────────────┐
              #       │   input_    │  (۸ فیچر خام)
              #       └──────┬──────┘
              #       │              │
              # (مسیر Wide)    (مسیر Deep)
              #       │              │
              #       │        ┌─────▼─────┐
              #       │        │  hidden1  │ (۳۰ نورون)
              #       │        └─────┬─────┘
              #       │              │
              #       │        ┌─────▼─────┐
              #       │        │  hidden2  │ (۳۰ نورون)
              #       │        └─────┬─────┘
              #       │              │
              #       └──────┬───────┘
              #       ┌───────▼───────┐
              #       │    concat     │ (۳۸ = ۸+۳۰)
              #       └───────┬───────┘
              #       ┌───────▼───────┐
              #       │    output     │ (۱ نورون، خروجی نهایی)
              #       └───────────────┘
```

```python
from tensorflow import keras

input_ = keras.layers.Input(shape=X_train.shape[1:])
hidden1 = keras.layers.Dense(30, activation='relu')(input_)
hidden2 = keras.layers.Dense(30, activation='relu')(hidden1)
concat = keras.layers.Concatenate()([input_, hidden2])
output = keras.layers.Dense(1)(concat)

model = keras.Model(inputs=[input_], outputs=[output])
```

```python
model.compile(loss="mean_squared_error", optimizer="sgd")

history = model.fit(X_train, Y_train, epochs=20,
                     validation_data=(X_valid, Y_valid))

mse_test = model.evaluate(X_test, Y_test)
print("MSE روی داده تست:", mse_test)

X_new = X_test[:3]
Y_pred = model.predict(X_new)
print("پیش‌بینی‌ها:", Y_pred)
```

# **model concatenate with two input and one output**

```python
# ┌──────────────────┐              ┌──────────────────┐
# │     input_A       │              │     input_B       │
# │  (wide_input)      │              │  (deep_input)      │
# │    ۵ فیچر          │              │    ۶ فیچر          │
# └─────────┬─────────┘              └─────────┬─────────┘
#           │                                  │
#           │                            ┌─────▼─────┐
#           │                            │  hidden1  │
#           │                            │ (۳۰ نورون) │
#           │                            └─────┬─────┘
#           │                                  │
#           │                            ┌─────▼─────┐
#           │                            │  hidden2  │
#           │                            │ (۳۰ نورون) │
#           │                            └─────┬─────┘
#           │                                  │
#           └────────────────┬─────────────────┘
#                     ┌───────▼───────┐
#                     │     concat     │
#                     │ (۵ + ۳۰ = ۳۵)  │
#                     └───────┬───────┘
#                     ┌───────▼───────┐
#                     │     output     │
#                     │   (۱ نورون)     │
#                     └───────────────┘
```

```python
from tensorflow import keras

input_A = keras.layers.Input(shape=[5], name="wide_input")
input_B = keras.layers.Input(shape=[6], name="deep_input")
hidden1 = keras.layers.Dense(30, activation='relu')(input_B)
hidden2 = keras.layers.Dense(30, activation='relu')(hidden1)
concat = keras.layers.concatenate([input_A, hidden2])
output = keras.layers.Dense(1, name="output")(concat)

model = keras.Model(inputs=[input_A, input_B], outputs=[output])
```

# **divide data for two inputs**

```python
model.compile(loss="mse", optimizer=keras.optimizers.SGD(learning_rate=1e-3))

X_train_A,X_train_B = X_train[:,:5],X_train[:,2:]
X_valid_A,X_valid_B = X_valid[:,:5],X_valid[:,2:]
X_test_A,X_test_B = X_test[:,:5],X_test[:,2:]
X_new_A, X_new_B = X_test_A[:3], X_test_B[:3]

history= model.fit((X_train_A,X_train_B) , Y_train , epochs=20 , validation_data=((X_valid_A,X_valid_B),Y_valid))
```

```python
mse_test = model.evaluate((X_test_A,X_test_B),Y_test)
Y_pred = model.predict((X_new_A,X_new_B))
print("پیش‌بینی‌ها:", Y_pred)
```

# **model concatenate with two input and one output**

```python
# ┌──────────────────┐              ┌──────────────────┐
# │     input_A       │              │     input_B       │
# │  (wide_input)      │              │  (deep_input)      │
# │    ۵ فیچر          │              │    ۶ فیچر          │
# └─────────┬─────────┘              └─────────┬─────────┘
#           │                                  │
#           │                            ┌─────▼─────┐
#           │                            │  hidden1  │
#           │                            │ (۳۰ نورون) │
#           │                            └─────┬─────┘
#           │                                  │
#           │                            ┌─────▼─────┐
#           │                            │  hidden2  │
#           │                            │ (۳۰ نورون) │
#           │                            └──┬─────┬───┘
#           │                               │     │
#           │                               │     └──────────────┐
#           └────────────────┬──────────────┘                    │
#                     ┌───────▼───────┐                   ┌───────▼───────┐
#                     │     concat     │                   │  aux_output   │
#                     │  (۵+۳۰ = ۳۵)   │                   │   (۱ نورون)    │
#                     └───────┬───────┘                   └───────────────┘
#                     ┌───────▼───────┐                     خروجی کمکی
#                     │     output     │                   (مستقیم از hidden2)
#                     │   (۱ نورون)     │
#                     └───────────────┘
#                      خروجی اصلی
```

```python
from tensorflow import keras

input_A = keras.layers.Input(shape=[5], name="wide_input")
input_B = keras.layers.Input(shape=[6], name="deep_input")
hidden1 = keras.layers.Dense(30, activation='relu')(input_B)
hidden2 = keras.layers.Dense(30, activation='relu')(hidden1)
concat = keras.layers.concatenate([input_A, hidden2])
output = keras.layers.Dense(1, name="output")(concat)
aux_output = keras.layers.Dense(1,name="aux_output")(hidden2)


model = keras.Model(inputs=[input_A, input_B], outputs=[output , aux_output])
```

# **divide data for two inputs and two output**

```python
model.compile(loss=["mse" ,"mse"], loss_weights=[0.9,0.1] , optimizer=keras.optimizers.SGD(learning_rate=1e-3))

X_train_A,X_train_B = X_train[:,:5],X_train[:,2:]
X_valid_A,X_valid_B = X_valid[:,:5],X_valid[:,2:]
X_test_A,X_test_B = X_test[:,:5],X_test[:,2:]
X_new_A, X_new_B = X_test_A[:3], X_test_B[:3]

history= model.fit(
    [X_train_A,X_train_B ], [Y_train , Y_train] , epochs=20
    , validation_data=([X_valid_A,X_valid_B],[Y_valid , Y_valid])
    )
```

# **Model Evaluation**

```python
total_loss,main_loss,aux_loss = model.evaluate(
    [X_test_A,X_test_B],[Y_test,Y_test]
)
Y_pred_main , Y_pred_aux = model.predict([X_new_A,X_new_B])
```
